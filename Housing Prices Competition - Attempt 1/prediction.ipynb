```python
# Housing Prices Competition for Kaggle Learn Users - 
#       Predict future sale prices for each house found in the given data set (data/test.csv)
#        using the training data (data/train.csv) 


# prediction.ipynb (this file) : trains the pandas model and produces predictions, which is then written to submissions.csv 
# submissions.csv : is the predicted SalesPrice that the pandas model produces, labeled by their Id

# Note : This is my first ever attempt to train a model


# imports
import pandas as pd
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor


# read the training data 
train_file_path = 'data/train.csv'
train_data = pd.read_csv(train_file_path)

# find all criteria to include in training, excluding Id and Saleprice 
feature_cols = train_data.columns.drop(["Id", "SalePrice"])

# get data to train model on 
X = train_data[feature_cols]
# convert text data to binary 
X = pd.get_dummies(X)
# get SalePrice to train model on 
y = train_data["SalePrice"]

# split data up to train model 
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)

# create RandomForestRegressor
# (max_leaf_nodes was found in the commented out code below)
house_model = RandomForestRegressor(max_leaf_nodes=500, random_state=1)

# fit the model to training set
house_model.fit(train_X, train_y)

# read the test csv
test_data_path = 'data/test.csv'
test_data = pd.read_csv(test_data_path)

test_X = pd.get_dummies(test_data[feature_cols])

# add any missing training columns as 0, and keep the exact same column order
test_X = test_X.reindex(columns=X.columns, fill_value=0)

# predict the SalePrices for the test csv
pred = house_model.predict(test_X)

# format the predictions to display the Id and SalePrice
predictions = pd.DataFrame({
    "Id" : test_data["Id"], 
    "SalePrice": pred
})

# write the predictions to submission.csv
predictions.to_csv("submission.csv", index=False)




# Below is the code for testing what max_leaf_nodes will produce the smallest mae =========================
# (resuluted in an max_leaf_nodes of 500)

"""
# max leaf nodes to test
candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500, 1000]


# store smallest mae
smallest_mae = -1              # intialized to -1 
# store number of nodes that produce smallest mae
smallest_mae_nodes = -1        # intialized to -1 

# get_mae() takes the max leaf node, the training and testing value and calculates the mae
def get_mae(max_leaf_node, train_X, val_X, train_y, val_y):

    # create a model with given max_leaf_node
    house_model = RandomForestRegressor(max_leaf_nodes = max_leaf_node, random_state = 1)
    # fit the model to train.csv
    house_model.fit(train_X, train_y)
    # create predictions
    pred = house_model.predict(val_X)
    # calc the mae
    mae = mean_absolute_error(val_y, pred)

    # return the mae
    return mae
    

# iterate through each of the leaf amounts to find smallest mae
for leaf_nodes in candidate_max_leaf_nodes:
    
    # calc the mae for the current leaf_nodes
    mae = get_mae(leaf_nodes, train_X, val_X, train_y, val_y)

    # check if there currently is not smallest mae, or if current mae is smaller than previous smallest
    if smallest_mae == -1 or smallest_mae > mae : 
            # store the mae
            smallest_mae = mae
            # store the number of max leaf nodes used 
            smallest_mae_nodes = leaf_nodes


# display the amount of leaf nodes with least mae
print(f"Smallest MAE is {smallest_mae} with {smallest_mae_nodes} leaf nodes")
"""
```
